```
def reverse(a):
  result=0
  while a>0:
    ld=a%10
    result=((result*10)+ld)
    a=a//10
  return a==result

reverse(2345)
```

```
def rever(s):
  orr=s
  s = s[::-1]
  return orr == s

rever("madam")
```

```
def rever(s):
  revarse=""
  for ch in s:
    revarse = ch + revarse
  return s == revarse

rever("madam")
```

```
def amstrom(num):
  nod = len(str(num))
  total=0
  while num>0:
    ld = num%10
    total=total+(ld**nod)
    num = num//10
  return total

amstrom(234)
```

```
def divis(num):
  result=[]
  for i in range(1,num+1):
    if num%i==0:
      result.append(i)
  return result

divis(20)
```
