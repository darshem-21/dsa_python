```
def reverse(a):
  result=0
  while a>0:
    ld=a%10
    result=((result*10)+ld)
    a=a//10
  return a==result

reverse(2345)
```

```
def rever(s):
  revarse=""
  for ch in s:
    revarse = ch + revarse
  return s == revarse

rever("madam")
```

```
def amstrom(num):
  nod = len(str(num))
  total=0
  while num>0:
    ld = num%10
    total=total+(ld**nod)
    num = num//10
  return total

amstrom(234)
```

```
def divis(num):
  result=[]
  for i in range(1,num+1):
    if num%i==0:
      result.append(i)
  return result

divis(20)
```

frequency in dicstnory

```
def freq(num):
  dict={}
  for i in range(0,len(num)):
    if num[i] in dict:
      dict[num[i]]+=1
    else:
      dict[num[i]]=1

  return dict

freq([1,2,3,4,1,2,3,4,5,6,7,7,6,7])
```

```
def hash(num):
  dict={}
  n=len(num)
  for i in range(0,n):
    dict[num[i]]=dict.get(num[i],0)+1
  return dict

hash([1,2,3,4,4,3,5,7,3])
```

hashing

```
def hashing(n,m):
  result=[]
  for i in n:
    count=0
    for j in m:
      if j==i:
        count+=1
        result.append((i,count))
  return result

hashing([4,5,6],[1,2,3,4,4,5,6])
```
