hashing q3

```python
def hash(n,m):
  dict={}
  for i in n:
    count=0
    for j in m:
      if i==j
        count+=1
    dict.append(i,count)
  return dict

hash([5,3,2,2,1,5,5][10,111,1,9,5,67,2])
```
