hashing q3

```python
def hash(n,m):
  d={}
  for i in n:
    count=0
    for j in m:
      if i==j:
        count+=1
    d[i]=count
  return d

hash([10,111,1,9,5,67,2], [5,3,2,2,1,5,5])
```

```python
def hash(n,m):
  result={}
  for ch in n:
    count = 0
    for j in m:
      if ch==j:
        count+=1
    result[ch]=count
  return result

hash("azwertyu",["d","q","w","w","e"])
```

longest contigious sub arrayl


```python
def longest_subarr(arr,l):
    max_len=0
    n=len(arr)
    for i in range(n):
      for j in range(i,n):
        valid = True
        for k in range(i,j):
          if abs(arr[k + 1]-arr[k])>l:
            valid = False
            break
        if valid:
          max_len=max(max_len,j-i+1);
        else:
          break

    return max_len

arr=[3,4,5,6,0,1,2]
l= 4
print(longest_subarr(arr,l))
```
